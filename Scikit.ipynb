```python
# Import necessary libraries
import skimage
import skimage.io
import skimage.filters
import skimage.segmentation
import skimage.morphology
import skimage.color
import skimage.restoration
import skimage.feature
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Example 1: Edge Detection (from the 'Edges' category)
# Why I chose this example: Edge detection is fundamental in image analysis, and I wanted to see how it performs on geological features.
# What I learned: I gained insights into the different edge detection algorithms and their parameters.

# Load your image
image1 = skimage.io.imread('/content/Gniess_outcrop.jpg')
# Apply the Sobel operator for edge detection
edges = skimage.filters.sobel(skimage.color.rgb2gray(image1))

# Display the original and edge-detected images
fig, axes = plt.subplots(ncols=2, sharex=True, sharey=True,
                         figsize=(8, 4))

axes[0].imshow(image1, cmap=plt.cm.gray)
axes[0].set_title('Original Image')

axes[1].imshow(edges, cmap=plt.cm.gray)
axes[1].set_title('Edge Detection (Sobel)')

plt.show()
```

```python
# Example 2: Segmentation (from the 'Segmentation & Morphology' category)
# Why I chose this example: Segmentation helps identify distinct regions in images, useful for tasks like mineral identification.
# What I learned: I learned how to apply segmentation techniques and how to fine-tune them for specific image features.

# Load your image
image2 = skimage.io.imread('/content/Thinsection_2.jpg')
# Perform SLIC segmentation
segments = skimage.segmentation.slic(image2, n_segments=100, compactness=10)

# Display the original and segmented images
fig, axes = plt.subplots(ncols=2, sharex=True, sharey=True,
                         figsize=(8, 4))

axes[0].imshow(image2)
axes[0].set_title('Original Image')

axes[1].imshow(skimage.color.label2rgb(segments, image2, kind='avg'))
axes[1].set_title('Segmented Image (SLIC)')

plt.show()
```

```python
# Example 3: Denoising (from the 'Restoration' category)
# Why I chose this example: Denoising can improve image quality, especially in satellite imagery with noise.
# What I learned: I learned about different noise models and denoising algorithms.

# Load your image
image3 = skimage.io.imread('/content/Pyrite_photo.jpg')
# Apply Total Variation denoising
denoised_image = skimage.restoration.denoise_tv_chambolle(image3, weight=0.1)

# Display the original and denoised images
fig, axes = plt.subplots(ncols=2, sharex=True, sharey=True,
                         figsize=(8, 4))

axes[0].imshow(image3, cmap=plt.cm.gray)
axes[0].set_title('Original Image')

axes[1].imshow(denoised_image, cmap=plt.cm.gray)
axes[1].set_title('Denoised Image (TV)')

plt.show()
```

```python
# Example 4: Feature Extraction (from the 'Feature Extraction' category)
# Why I chose this example: Feature extraction helps identify key image characteristics, like texture, that can be used for classification.
# What I learned: I explored texture analysis methods and how to apply them to images of rocks.

# Load your image
image4 = skimage.io.imread('/content/Quarry_photo.jpg')
# Extract LBP features
lbp_features = skimage.feature.local_binary_pattern(skimage.color.rgb2gray(image4), P=8, R=1)

# Display the original image and LBP features
fig, axes = plt.subplots(ncols=2, sharex=True, sharey=True,
                         figsize=(8, 4))

axes[0].imshow(image4, cmap=plt.cm.gray)
axes[0].set_title('Original Image')

axes[1].imshow(lbp_features, cmap=plt.cm.gray)
axes[1].set_title('LBP Features')

plt.show()
```

```python
# Example 5: Color Manipulation (from the 'Color' category)
# Why I chose this example: Color manipulation can be useful for highlighting specific features or improving image visualization.
# What I learned: I learned about different color spaces and how to manipulate them.

# Load your image
image5 = skimage.io.imread('/content/Thinsection_1.jpg')
# Convert the image from RGB to HSV
hsv_image = skimage.color.rgb2hsv(image5)

# Modify the hue channel (e.g., increase saturation)
hsv_image[:, :, 1] = hsv_image[:, :, 1] * 1.5

# Convert back to RGB
modified_image = skimage.color.hsv2rgb(hsv_image)

# Display the original and modified images
fig, axes = plt.subplots(ncols=2, sharex=True, sharey=True,
                         figsize=(8, 4))

axes[0].imshow(image5)
axes[0].set_title('Original Image')

axes[1].imshow(modified_image)
axes[1].set_title('Modified Image (HSV)')

plt.show()
```
